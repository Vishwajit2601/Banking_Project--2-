### Read Raw CSV Data into DataFrames

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

credential = "banking_storage_credential"
storage = "bankingstorage26"
```

```python
# Read all CSV files from raw container
raw_path = f"abfss://raw@{storage}.dfs.core.windows.net/csv/"
print(raw_path)

# Read Accoounts Dataset
account_path = f"{raw_path}accounts.csv"
accounts_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(account_path)

# Read atm_transaction Dataset
atm_path = f"{raw_path}atm_transactions.csv"
atm_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(atm_path)

# Read branches Dataset
branches_path = f"{raw_path}branches.csv"
branches_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(branches_path)

# Read credit_cards Dataset
credit_cards_path = f"{raw_path}credit_cards.csv"
credit_cards_df = spark.read.option("header", True)\
                    .option("inferSchema", True)\
                    .csv(credit_cards_path)

# Read customers Dataset
customers_path = f"{raw_path}customers.csv"
customers_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(customers_path)

# Read employees Dataset
employees_path = f"{raw_path}employees.csv"
employees_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(employees_path)

# Read fraud_alerts Dataset
fraud_path = f"{raw_path}fraud_alerts.csv"
fraud_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(fraud_path)

# Read kyc_documents Dataset
kyc_path = f"{raw_path}kyc_documents.csv"
kyc_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(kyc_path)

# Read loans Dataset
loans_path = f"{raw_path}loans.csv"
loans_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(loans_path)

# Read transactions Dataset
transactions_path = f"{raw_path}transactions.csv"
transactions_df = spark.read.option("header", True)\
                .option("inferSchema", True)\
                .csv(transactions_path)

print("✅All CSV files loaded successfully")
```

### 4: Write Raw Data to Bronze Delta Tables

```python
# Write each DataFrame as Delta table in Bronze layer
bronze_path = f"abfss://bronze@{storage}.dfs.core.windows.net/"

# Accounts
accounts_df.write.format("delta").mode("overwrite").save(f"{bronze_path}accounts")

# ATM Transactions
atm_df.write.format("delta").mode("overwrite").save(f"{bronze_path}atm_transactions")

# Branches
branches_df.write.format("delta").mode("overwrite").save(f"{bronze_path}branches")

# Credit Cards
credit_cards_df.write.format("delta").mode("overwrite").save(f"{bronze_path}credit_cards")

# Customers
customers_df.write.format("delta").mode("overwrite").save(f"{bronze_path}customers")

# Employees
employees_df.write.format("delta").mode("overwrite").save(f"{bronze_path}employees")

# Fraud Alerts
fraud_df.write.format("delta").mode("overwrite").save(f"{bronze_path}fraud_alerts")

# KYC Documents
kyc_df.write.format("delta").mode("overwrite").save(f"{bronze_path}kyc_documents")

# Loans
loans_df.write.format("delta").mode("overwrite").save(f"{bronze_path}loans")

# Transactions
transactions_df.write.format("delta").mode("overwrite").save(f"{bronze_path}transactions")

print("✅ All tables written to Bronze layer!")
```

### 5: Create Bronze Tables in Unity Catalog

```python
%sql
-- Create external tables pointing to Bronze Delta files
CREATE TABLE IF NOT EXISTS banking_catalog.bronze.accounts
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/accounts/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.atm_transactions
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/atm_transactions/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.branches
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/branches/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.credit_cards
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/credit_cards/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.customers
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/customers/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.employees
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/employees/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.fraud_alerts
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/fraud_alerts/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.kyc_documents
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/kyc_documents/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.loans
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/loans/';

CREATE TABLE IF NOT EXISTS banking_catalog.bronze.transactions
USING DELTA
LOCATION 'abfss://bronze@bankingstorage26.dfs.core.windows.net/transactions/';

```
